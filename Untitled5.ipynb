```python
import os, shutil, random
from pathlib import Path

SRC = Path("/content/dataset-resized")
DST = Path("/content/dataset")
IMG_TRAIN = DST/"images"/"train"
IMG_VAL   = DST/"images"/"val"
LBL_TRAIN = DST/"labels"/"train"
LBL_VAL   = DST/"labels"/"val"

for d in [IMG_TRAIN, IMG_VAL, LBL_TRAIN, LBL_VAL]:
    d.mkdir(parents=True, exist_ok=True)

trash_imgs = list((SRC/"trash").glob("*.*"))
other_imgs = []
for c in ["glass","paper","plastic","cardboard","metal"]:
    other_imgs += list((SRC/c).glob("*.*"))

R = random.Random(42)

def write_labels(img_list, is_trash, train=True):
    for p in img_list:
        if train:
            shutil.copy2(p, IMG_TRAIN/p.name)
            lbl = LBL_TRAIN/(p.stem + ".txt")
        else:
            shutil.copy2(p, IMG_VAL/p.name)
            lbl = LBL_VAL/(p.stem + ".txt")

        if is_trash:
            lbl.write_text("0 0.5 0.5 1 1\n")
        else:
            lbl.write_text("")

def split_and_copy(imgs, is_trash):
    R.shuffle(imgs)
    s = int(0.8 * len(imgs))
    write_labels(imgs[:s], is_trash, True)
    write_labels(imgs[s:], is_trash, False)

split_and_copy(trash_imgs, True)
split_and_copy(other_imgs, False)

print("Dataset prepared successfully!")
```

```python
yaml = """
names: ['trash']
nc: 1
train: /content/dataset/images/train
val: /content/dataset/images/val
"""

with open("/content/data.yaml","w") as f:
    f.write(yaml)

print("data.yaml created!")
```

```python
from ultralytics import YOLO
import datetime, os

STUDENT_ID = "rishi_rob05"
EXP = f"trash_{STUDENT_ID}"

model = YOLO("yolov8n.yaml")  # <-- architecture only, NO .pt load → NO unpickling issues

results = model.train(
    data="/content/data.yaml",
    epochs=10,     # increase if time allows
    imgsz=320,
    batch=8,
    name=EXP,
    device="cpu"  # automatically selects CPU; change to 0 if GPU available
)

print("Training complete!")
print("Best model --> /content/runs/detect/{}/weights/best.pt".format(EXP))
```
